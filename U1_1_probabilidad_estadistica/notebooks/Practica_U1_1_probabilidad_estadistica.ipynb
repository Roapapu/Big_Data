## Actividad 3

```python
# Seleccionar año histórico y columna de cambio anual de población
año_seleccionado = 2023
columna_cambio = 'Annual change in population'

# Filtrar el DataFrame
df_filtrado = df[df['Year'] == año_seleccionado].copy()

# Calcular medidas estadísticas descriptivas básicas
media = df_filtrado[columna_cambio].mean()
mediana = df_filtrado[columna_cambio].median()
varianza = df_filtrado[columna_cambio].var()
desviacion_std = df_filtrado[columna_cambio].std()
valor_min = df_filtrado[columna_cambio].min()
valor_max = df_filtrado[columna_cambio].max()

# Identificar las entidades correspondientes al mínimo y máximo
entidad_min = df_filtrado.loc[df_filtrado[columna_cambio].idxmin(), 'Entity']
entidad_max = df_filtrado.loc[df_filtrado[columna_cambio].idxmax(), 'Entity']

# Mostrar resultados
print(f"--- Estadísticas para el año {año_seleccionado} ---")
print(f"Media: {media}")
print(f"Mediana: {mediana}")
print(f"Varianza: {varianza}")
print(f"Desviación estándar: {desviacion_std}")
print(f"Mínimo: {valor_min} (Entidad: {entidad_min})")
print(f"Máximo: {valor_max} (Entidad: {entidad_max})")
```
